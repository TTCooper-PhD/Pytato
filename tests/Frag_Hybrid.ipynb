```python
import pandas as pd
```

```python
df = pd.read_csv('Test_Hybrid.txt', sep='\t')
```

```python
df.head(20)
```

```python
df.shape
```

```python
import re
# Define the allowed modification masses
allowed_masses = {"42.0106", "15.9949", "57.0214"}

# Function to check for any unexpected modification
def has_unexpected_mod(mod_string):
    matches = re.findall(r'\(([^)]+)\)', mod_string)
    return any(m not in allowed_masses for m in matches)

# Filter the dataframe
df_filtered = df[df['C: Assigned Modifications'].apply(has_unexpected_mod)]

# Show result
df_filtered.shape
```

# Rename columns to remove prefix up to and including "_Glycoproteomics_"



```python
df_filtered.columns = df_filtered.columns.str.replace(r'.*_Glycoproteomics_', '', regex=True)
```
